## 2. Cac kieu du lieu co ban va thuat toan tren chung

### 2.1 String (chuoi)
La kieu du lieu de luu van ban, dat trong dau nhay don hoac kep. Hay dung khi can luu ten, thong diep, ...

```python
ho_ten = "Tran Thi B"

loi_chao = "Xin chao, " + ho_ten
print(loi_chao)

print(ho_ten[0:4])
print(ho_ten.upper())
print(ho_ten.lower())
print(len(ho_ten))
print(ho_ten.replace("B", "C"))
```

### 2.2 Integer (so nguyen)
Kieu so khong co phan thap phan, hay dung de dem, luu tuoi, so luong...

```python
a = 15
b = 4

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

### 2.3 Float (so thuc)
Co phan thap phan, dung khi tinh toan can chinh xac nhu gia tien, diem so.

```python
x = 5.75
y = 2.5

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(round(x, 1))
print(int(x))
```

### 2.4 Bool (logic)
Chi co 2 gia tri True hoac False, dung trong cac dieu kien, so sanh.

```python
p = True
q = False

print(p and q)
print(p or q)
print(not p)

print(10 > 5)
print(10 == 5)
```

## 3. Cac kieu du lieu nang cao: List, Tuple, Set, Dict

### 3.1 List (danh sach)
La tap hop co thu tu, co the them/xoa/sua duoc, cho phep trung gia tri. Hay dung de luu danh sach diem, danh sach san pham...

```python
diem = [8.5, 7.0, 9.0, 6.5]
print(diem)

diem.append(10.0)
print(diem)

diem[0] = 9.5
print(diem)

diem.remove(6.5)
print(diem)

print("Trung binh:", sum(diem) / len(diem))
```

### 3.2 Tuple (bo)
Cung co thu tu nhung khong sua duoc sau khi tao. Hay dung cho du lieu co dinh nhu toa do.

```python
toa_do = (10, 20)
print(toa_do)
print("x =", toa_do[0], ", y =", toa_do[1])

# toa_do[0] = 5  -> loi vi tuple khong sua duoc

sv = ("Le Van C", 21, "Cong nghe thong tin")
print(sv)
```

### 3.3 Set (tap hop)
Khong co thu tu va khong duoc trung phan tu. Hay dung khi can loai bo gia tri trung.

```python
mon_a = {"Toan", "Ly", "Hoa", "Van"}
mon_b = {"Toan", "Sinh", "Anh"}

print(mon_a)

mon_a.add("Anh")
print(mon_a)

print("Giao:", mon_a & mon_b)
print("Hop:", mon_a | mon_b)
print("Hieu:", mon_a - mon_b)
```

### 3.4 Dictionary (tu dien)
Luu theo cap key: value, key khong duoc trung. Hay dung cho du lieu co cau truc nhu ho so sinh vien.

```python
sv = {
    "ho_ten": "Pham Thi D",
    "tuoi": 20,
    "nganh": "Khoa hoc may tinh"
}

print(sv)
print(sv["ho_ten"])

sv["tuoi"] = 21
sv["diem_tb"] = 8.2
print(sv)

for key, value in sv.items():
    print(key, ":", value)
```

## 4. Cau truc dieu khien trong Python

### 4.a. Dieu kien if - elif - else
Dung de re nhanh chuong trinh tuy theo dieu kien dung hay sai.

```python
diem = 7.5

if diem >= 8.5:
    xep_loai = "Gioi"
elif diem >= 7.0:
    xep_loai = "Kha"
elif diem >= 5.0:
    xep_loai = "Trung binh"
else:
    xep_loai = "Yeu"

print("Diem", diem, "xep loai:", xep_loai)
```

### 4.b. Vong lap for va while
Dung de lap lai mot doan lenh nhieu lan.

```python
for i in range(1, 6):
    print("Lan lap thu", i)

print("---")

dem = 0
while dem < 5:
    print("Dem:", dem)
    dem += 1
```

## 5. Khai bao ham
Dung tu khoa def de dinh nghia ham, giup tai su dung code.

```python
def tinh_trung_binh(ds):
    return sum(ds) / len(ds)

def xep_loai(diem):
    if diem >= 8.5:
        return "Gioi"
    elif diem >= 7.0:
        return "Kha"
    elif diem >= 5.0:
        return "Trung binh"
    else:
        return "Yeu"

diem_so = [8.0, 7.5, 9.0, 6.0]
tb = tinh_trung_binh(diem_so)
print("Diem trung binh:", tb)
print("Xep loai:", xep_loai(tb))
```

## 6. Doc ghi file van ban
Dung ham open() voi cac che do: w la ghi (ghi de), a la ghi them, r la doc.

```python
with open("vidu.txt", "w", encoding="utf-8") as f:
    f.write("Xin chao cac ban!\n")
    f.write("Day la dong thu hai.\n")
    f.write("Python rat thu vi.\n")

print("Da ghi file xong")
```

```python
with open("vidu.txt", "r", encoding="utf-8") as f:
    noi_dung = f.read()
print(noi_dung)
```

```python
with open("vidu.txt", "r", encoding="utf-8") as f:
    for so_dong, dong in enumerate(f, start=1):
        print("Dong", so_dong, ":", dong.strip())
```

```python
with open("vidu.txt", "a", encoding="utf-8") as f:
    f.write("Dong nay them vao sau.\n")

with open("vidu.txt", "r", encoding="utf-8") as f:
    print(f.read())
```
